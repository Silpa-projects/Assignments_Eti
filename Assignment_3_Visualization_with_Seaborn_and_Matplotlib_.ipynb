```python
# Assignment - 3
# Topic - Visualization with Seaborn and Matplotlib

#Import Libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load Titanic Dataset

titanic = sns.load_dataset("titanic")
```

Question 1'

Countplot: Passenger Count by Embarked Port

```python
plt.figure(figsize=(8, 5))
```

```python
sns.countplot(
    data=titanic,
    x="embarked",
    hue="sex"
)
```

```python
plt.title("Passenger Count by Embarked Port")
plt.xlabel("Embarked Port")
plt.ylabel("Number of Passengers")
plt.legend(title="Sex")
plt.show()
```

Question 2

Swarmplot: Age Distribution across Classes



```python
plt.figure(figsize=(9,6))
```

```python
sns.swarmplot(
    data=titanic,
    x="pclass",
    y="age",
    hue="sex"
)
```

```python
plt.title("Age Distribution Across Classes")
plt.xlabel("Passenger Class")
plt.ylabel("Age")
plt.show()
```

Question 3

Pie Chart: Gender Distribution

```python
gender_count = titanic["sex"].value_counts()
plt.figure(figsize=(6, 6))
```

```python
plt.pie(
    gender_count,
    labels=gender_count.index,
    autopct="%1.1f%%",
    explode=(0.05,0),
    startangle=90
)
```

```python
plt.title("Gender Distribution")
plt.show()
```

Question 4

Heatmap: Correlation Matrix

```python
corr=titanic[['age','fare','pclass','sibsp','parch','survived']].corr()
```

```python
plt.figure(figsize=(8, 6))
```

```python
sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm"
)
```

```python
plt.title("Correlation Matrix")
plt.show()
```

Question 5

Violin Plot: Fare by Class and Gender

```python
plt.figure(figsize=(8, 6))
sns.violinplot(
    data=titanic,
    x="pclass",
    y="fare",
    hue="sex"
)
```

```python
plt.title("Fare distribution by Class and Gender")
plt.xlabel("Passenger Class")
plt.ylabel("Fare")
plt.show()
```

Question 6

Scatter Plot: Age vs Fare

```python
plt.figure(figsize=(8, 6))
sns.scatterplot(
    data=titanic,
    x="age",
    y="fare",
    hue="survived"
)
```

```python
plt.title("Age vs Fare:")
plt.xlabel("Age")
plt.ylabel("Fare")
plt.show()
```
