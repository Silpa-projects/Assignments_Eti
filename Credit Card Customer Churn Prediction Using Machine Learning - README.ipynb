## Executive Summary

This project develops a machine learning classification system to predict whether a credit-card customer is likely to churn. The Kaggle Credit Card Customers / BankChurners dataset contains 10,127 records and 23 original columns. After removing CLIENTNUM and the two Naive Bayes classifier columns, 20 columns remained, including the target Attrition_Flag and 19 input features.
The workflow covered data cleaning, exploratory data analysis, feature engineering, train-test splitting, standardization, feature selection, comparison of five classification algorithms, model evaluation, hyperparameter tuning, model saving, and prediction on unseen customer records.
The tuned Gradient Boosting Classifier was selected as the final model. On the held-out test set it achieved 96.79% accuracy, 95.14% precision, 84.31% recall, 89.40% F1-score, and 99.17% ROC-AUC.


## 1. Overview of Problem Statement

Customer attrition is a major challenge for banks because retaining existing customers is important. Identifying customers who are likely to leave can help a bank take preventive action. Machine learning can identify patterns associated with customer attrition and support customer-retention decisions.

## 2. Objective

To develop and evaluate machine learning classification models that can predict whether a credit-card customer is likely to churn based on demographic, financial, account, and transaction-related characteristics.

## 5. Data Preprocessing and Data Cleaning

•	Original dataset: 10,127 rows × 23 columns.
•	CLIENTNUM was removed because it is an identifier rather than a predictive customer-behaviour feature.
•	The two Naive Bayes classifier columns were removed as specified during dataset preparation.
•	Final cleaned dataset: 10,127 rows × 20 columns.
•	Total missing values: 0.
•	Duplicate records: 0.
•	Unknown categorical values were retained as categorical levels rather than being treated as missing.
•	Statistical outliers were inspected and retained where they could represent genuine financial or transaction behaviour.

## 6. Exploratory Data Analysis

The EDA examined target distribution, categorical distributions, numerical distributions, relationships between variables, correlations, and churn patterns using the required visualization types.
•	Existing Customers: 8,500 (83.93%).
•	Attrited Customers: 1,627 (16.07%).
•	The target is therefore imbalanced.
•	Average Total_Trans_Amt was lower for Attrited Customers (about 3,095) than Existing Customers (about 4,655).
•	Contacts_Count_12_mon and Months_Inactive_12_mon showed positive association with attrition.
•	Total_Trans_Ct, Total_Ct_Chng_Q4_Q1, Total_Revolving_Bal and Avg_Utilization_Ratio showed negative association with attrition.
These relationships are associations observed in the dataset and should not be interpreted as causal effects.

## 16. Interpretation and Conclusion

The project successfully developed a customer churn prediction system. The tuned Gradient Boosting model provided strong performance across the main classification metrics. Its high precision indicates that a large proportion of customers predicted as attrited were correctly classified, while its recall indicates that it identified a substantial proportion of the actual attrited customers.
The model can support customer-retention strategies by helping identify customers who may be at higher risk of leaving. Predictions should be used as decision-support information rather than as proof of causal relationships.

## 17. Limitations

•	The dataset represents a specific credit-card customer population and may not represent every bank or market.
•	The target variable is imbalanced, with many more existing customers than attrited customers.
•	The unseen-data demonstration uses only three new records and is illustrative rather than a large independent validation study.
•	Customer behaviour can change over time, so model performance should be monitored and the model retrained when appropriate.
