```python
import pandas as pd
import numpy as np
```

```python
# Load the dataset
df = pd.read_csv("C:\\Users\\silpa\\Downloads\\winequality-red.csv", sep=";")
```

```python
# Display first 5 rows
print("First 5 rows of the dataset:")
display(df.head())
```

```python
# Check for missing values
print("Missing values before treatment:")
print(df.isnull().sum())
```

```python
# Replace missing values with column mean
df = df.fillna(df.mean(numeric_only=True))
```

```python
# Check for missing values after treatment
print("Missing values after treatment:")
print(df.isnull().sum())
```

```python
# Extract alcohol vector
alcohol_vector = df["alcohol"].values
```

```python
# Extract citric acid vector
citric_acid_vector = df["citric acid"].values
```

```python
print("Alcohol Vector:")
print(alcohol_vector)
```

```python
print("Citric Acid Vector:")
print(citric_acid_vector)
```

```python
np.cov
```

```python
# Select alcohol and density
X = df[["alcohol", "density"]].values
```

```python
# Calculate covariance matrix
covariance_matrix = np.cov(X.T)
```

```python
print("Covariance Matrix:")
print(covariance_matrix)
```

```python
# Eigen decomposition
eigenvalues, eigenvectors = np.linalg.eig(covariance_matrix)
```

```python
print("Eigenvalues:")
print(eigenvalues)
```

```python
print("\nEigenvectors:")
print(eigenvectors)
```

```python
# Sort eigenvalues in descending order
idx = np.argsort(eigenvalues)[::-1]
```

```python
sorted_eigenvalues = eigenvalues[idx]
sorted_eigenvectors = eigenvectors[:, idx]
```

```python
print("Eigenvalues in descending order:")
print(sorted_eigenvalues)
```

```python
print("Corresponding Eigenvectors:")
print(sorted_eigenvectors)
```
